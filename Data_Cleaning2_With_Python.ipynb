```python
#Import Library
import pandas as pd
import numpy as np
import random

#Buat Data Kotor
random.seed(42)

names = [
    "Andi","Budi","Citra","Dewi","Eko","Farhan","Gita","Hendra",
    "Indah","Joko","Kiki","Lina","Maya","Nina","Oscar","Putri",
    "Rina","Sinta","Tono","Wawan"
]

cities = [
    "Jakarta","Bandung","Surabaya","Semarang","Medan",
    "Yogyakarta","Malang","Denpasar"
]

gender = [
    "M","Male","L","Laki-laki",
    "F","Female","P","Perempuan"
]

rows = []

for i in range(100):

    name = random.choice(names)

    # Nama dibuat tidak konsisten
    if random.random() < 0.2:
        name = name.upper()
    elif random.random() < 0.2:
        name = name.lower()

    if random.random() < 0.15:
        name = " " + name + " "

    age = random.choice([
        random.randint(20,55),
        "",
        None,
        -5,
        150
    ])

    city = random.choice(cities)

    if random.random() < 0.2:
        city = city.lower()

    salary = random.choice([
        random.randint(4000000,15000000),
        f"Rp {random.randint(4,15)}.000.000",
        f"{random.randint(4,15)},000,000",
        "",
        None
    ])

    join_date = random.choice([
        "2024-01-10",
        "10/01/2024",
        "Jan 10 2024",
        "2024/01/10",
        ""
    ])

    gender_value = random.choice(gender)

    rows.append([
        i+1,
        name,
        age,
        gender_value,
        city,
        salary,
        join_date
    ])

# Tambahkan duplikasi
rows.append(rows[5])
rows.append(rows[20])
rows.append(rows[35])

df = pd.DataFrame(rows, columns=[
    "ID",
    "Name",
    "Age",
    "Gender",
    "City",
    "Salary",
    "JoinDate"
])

#hasil Data Kotor
print(df)

df.to_csv("dirty_employee_data.csv", index=False)
```

```python
#Handle standardisasi Type Data

#Age
df['Age']=df['Age'].replace('',np.nan)
df['Age']=df['Age'].replace(-5,np.nan)
df['Age']=df['Age'].astype('Int64')

#Salary
df['Salary']=df['Salary'].astype(str)
df['Salary']=df['Salary'].replace('',np.nan)
df['Salary']=df['Salary'].str.replace('Rp','')
df['Salary']=df['Salary'].str.replace('.','', regex=False)
df['Salary']=df['Salary'].str.replace(',','', regex=False)
df['Salary']=df['Salary'].str.replace(' ','', regex=False)
df['Salary']=df['Salary'].str.replace('-','', regex=False)

# Use pd.to_numeric with errors='coerce' to convert non-numeric values (including '') to NaN
df['Salary']=pd.to_numeric(df['Salary'],errors='coerce')
df['Salary']=df['Salary'].astype('Int64')

#JoinDate
df['JoinDate']=df['JoinDate'].astype(str)
df['JoinDate']=df['JoinDate'].replace('',np.nan)
df['JoinDate']=df['JoinDate'].str.replace('/','-')
df['JoinDate']=df['JoinDate'].str.replace('Jan','01')
df['JoinDate']=df['JoinDate'].str.replace(' ','-')
df['JoinDate']=pd.to_datetime(df['JoinDate'],errors='coerce',dayfirst=True)
```

```python
#Handle Missing Value
df['Age']=df['Age'].fillna(df['Age'].mean().astype(int))
df['Salary']=df['Salary'].fillna(df['Salary'].mean().astype(int))
df.info()
```

```python
#Handle Missing Value JoinDate dengan Modus
df['JoinDate']=df['JoinDate'].fillna(df['JoinDate'].mode()[0])
```

```python
#Handle kolom Name, Gender, City
df['Name']=df['Name'].str.lower().str.strip().str.title()
df['Gender']=df['Gender'].str.lower().str.strip().map({'m':'Laki-Laki','l':'Laki-Laki','male':'Laki-Laki',
                                                       'f':'Perempuan','p':'Perempuan','female':'Perempuan',
                                                       'laki-laki':'Laki-Laki','perempuan':'Perempuan'})
df['City']=df['City'].str.lower().str.strip().str.title()
df
```
